## 1) Qualitative(categorical)DATA

### data that describe qualities or characteristics, and cannot be measured numerically,that data is known as categorical or qualitative data
a) Nominal data:-categories with no inherent order

```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
nominal_data=pd.Series(['Red','Blue','Green','Red','Blue','Red','Green','Red','Blue','Red'],name='Colors')

print(nominal_data)

print(nominal_data.value_counts())
```

```python
#visulization
nominal_data.value_counts().plot(kind='bar',color=['red','blue','green'],title='Nominal Data Visualization')
plt.title('Nominal Data Visualization',fontsize=16)
plt.xlabel('Colors',fontsize=14)
plt.ylabel('Count',fontsize=14)
plt.show()
```

## Ordinal DATA:-Categories with a meanining full order/ranking

```python
ordinal_data=pd.Series(['Low','Medium','High','Low','Medium','Low','High','Low','Medium','Low'],name='Satisfaction')
#define order of categories
order=['Low','Medium','High']
ordinal_data=pd.Categorical(ordinal_data, categories=order, ordered=True)
print(ordinal_data.value_counts().sort_index())
```

```python
# VISUALIZATION
ordinal_data.value_counts().sort_index().plot(kind='bar',color=['green','yellow','red'],title='Ordinal Data Visualization')
plt.title('Ordinal Data Visualization',fontsize=16)
plt.show()
```

# 2. Quntitative(Numerical)Data:

## Data that can be measured and expressed numerically that is known as quantitative data or numerical data
a) Discrete Data :- Countable values(Whole Number),often counts

```python
import numpy as np
import seaborn as sns
```

```python
np.random.seed(42)
discrete_data=pd.Series(np.random.randint(0,6,100),name='Children')
print(discrete_data)
print(discrete_data.value_counts().sort_index())
```

```python
sns.countplot(x=discrete_data)
plt.title('Discrete Data -number of children')
plt.show()
```

## b) Continuous Data:- Any value within a range can be considered as continuous data.(can be decimal/fractional)

```python
continuous_data=pd.Series(np.random.normal(170,10,1000),name='Height')
print(continuous_data.describe())
print(continuous_data)
```

```python
sns.histplot(continuous_data,kde=True,color='purple')
plt.title('Continuous Data - Height Distribution')
plt.xlabel('height(cm)')
plt.show()
```

```python
mean=continuous_data.mean()
median=continuous_data.median()
mode=continuous_data.mode()[0]
sns.histplot(continuous_data,kde=True,color='gray')
plt.axvline(mean,color='red',linestyle='--',linewidth=2,label=f'Mean={mean:.2f}')
plt.axvline(median,color='blue',linestyle='--',linewidth=2,label=f'Median={median:.2f}')
plt.axvline(mode,color='green',linestyle='--',linewidth=2,label=f'Mode={mode:.2f}')
plt.title('Continuous Data - Height Distribution')
plt.xlabel('height(cm)')
plt.show()
```

## Normal distribution :- A distribution shows how data values sre spread accross a range,the KDE(Kernel Density Estimatie) is a sooth curve that estimate the probability of the data like a smoothed histogram

## <font color='pink'> The Three Key Distribution are ,

1. Normal(symmetric)(Bell hSaped)blanced both side
2. Right skewed(negative)long tail on right side mean>median
3. left skewed(negetive)long tail on left side mean<median





```python
np.random
n=10000
normal_data=np.random.normal(loc=50,scale=10,size=n)
right_skewed=np.random.exponential(scale=10,size=n)+20
left_skewed=20-np.random.exponential(scale=10,size=n)
df=pd.DataFrame({'Normal':normal_data,'Right Skewed':right_skewed,'Left Skewed':left_skewed})
print(df)

```

```python
from scipy import stats
fig,axes=plt.subplots(nrows=1,ncols=3,figsize=(18,5))
datasets=[('Normal',normal_data,'steelblue'),
          ('Right Skewed',right_skewed,'salmon'),
          ('Left Skewed',left_skewed,'mediumseagreen')]
for ax,(title,data,color) in zip(axes,datasets):
    sns.histplot(data,kde=True,ax=ax,color=color,stat='density',bins=40,alpha=0.5)
    mean_val=np.mean(data)
    median_val=np.median(data)
    ax.axvline(mean_val,color='red',linestyle='--',linewidth=2,label=f'Mean={mean_val:.1f}')
    ax.axvline(median_val,color='blue',linestyle='-',linewidth=2,label=f'Median={median_val:.1f}')
    skew=stats.skew(data)
    ax.set_title(f'{title} \nSkewness={skew:.2f}',fontsize=14)
    ax.set_xlabel('Value',fontsize=12)
    ax.set_ylabel('Density',fontsize=12)
    ax.legend()
plt.show()
```
