```python
fig=plt.figure(figsize=(12,8))
fig.add_subplot(1,1,1)
sns.heatmap(cormat,annot=True,cmap='YlGnBu',linewidths=0.5)
plt.show()
```

```python
def getcorrelation(dataset, threshold):
    features = []
    values = []
    for i, index in enumerate(dataset.index):
        if abs(dataset.iloc[i]) > threshold:
            features.append(index)
            values.append(dataset[index])
    df = pd.DataFrame(data=values, index=features, columns=['corr_data'])
    return df
```

```python
threshold=0.50
corr_value=getcorrelation(cormat['medv'],threshold)
corr_value
```

```python
correlated_data=df[corr_value.index]
correlated_data.head()
```

```python
sns.pairplot(correlated_data)
```

```python
sns.heatmap(correlated_data.corr(),annot=True,cmap='YlGnBu',annot_kws={'size':10})
plt.show()
```

```python
correlated_data.columns
```

```python
x=correlated_data[['rm','ptratio','lstat']]
y=correlated_data['medv']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lin_reg=LinearRegression()
```

```python
lin_reg.fit(x_train,y_train)
```

```python
y_pred=lin_reg.predict(x_test)
```

```python
result_df=pd.DataFrame(data=[y_pred,y_test],index=['Predicted','Actual'])
result_df.T.head()
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
```

```python
print("The MSE is",mean_squared_error(y_test,y_pred))
print("The MAE is",mean_absolute_error(y_test,y_pred))
print("The R² is",r2_score(y_test,y_pred))
```
