## ndim: it returns the number of dimensions of an array.

```python
print(a.ndim)
print(b.ndim)
print(c.ndim)
```

## shape:it return a tuple consisting of array dimensions. it can also be used to resize the array.

```python
print(a.shape)
print(b.shape)
print(c.shape)
```

## len:- it return length of the array for ndimesion.

```python
print(len(a))
print(len(b))
print(len(c))
```

## dtype:- it return the data type of the elements in the array.

```python
print(a.dtype)
print(b.dtype)
print(c.dtype)
```

## np.linspace:this function is similar to arange()function.in this function insted of stepsize the number of evenly spaced values between the interval is specified
1) start:the starrting value of the sequence
2) stop:the end value of the sequence
3) num:the number of evenly spaced sample to be generated default is 50
4) retstep=if true returns samples and step between the consecutive number

```python
new_g=np.linspace(1,5,retstep=True)
print(new_g)
print(len(new_g))
```

## Write A Program to generate 16 values using np.arange() and reshape them into different types of arrays such as 2×3, 3×4, and 4×5

```python
num = np.arange(1, 17)
print("Original array:")
print(num)
print()

a = num[:6].reshape(2, 3)
print("Array 1 (2x3):")
print(a)
print()

b = num[:12].reshape(3, 4)
print("Array 2 (3x4):")
print(b)
print()

num2 = np.arange(1, 21)
c = num2.reshape(4, 5)
print("Array 3 (4x5):")
print(c)
```
